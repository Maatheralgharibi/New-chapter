```python
import sys
sys.setrecursionlimit(10**7)

n, m = map(int, input().split())
board = [list(input().strip()) for _ in range(n)]

dist = [[-1]*m for _ in range(n)]
visited = [[0]*m for _ in range(n)]
dirs = {'<': (0, -1), '>': (0, 1), '^': (-1, 0), 'v': (1, 0)}

def dfs(i, j):
    if board[i][j] == '#':
        return 0
    if visited[i][j] == 1:
        print(-1)
        sys.exit()
    if visited[i][j] == 2:
        return dist[i][j]

    visited[i][j] = 1
    dx, dy = dirs[board[i][j]]
    ni, nj = i + dx, j + dy

    dist[i][j] = 1 + dfs(ni, nj)
    visited[i][j] = 2
    return dist[i][j]
lengths = []
for i in range(n):
    for j in range(m):
        if board[i][j] != '#':
            lengths.append(dfs(i, j))
lengths.sort(reverse=True)
if len(lengths) == 0:
    print(0)
elif len(lengths) == 1:
    print(lengths[0] * 2)
else:
    print(lengths[0] + lengths[1])
```
