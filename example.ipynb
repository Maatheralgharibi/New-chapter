### D. Powerful Ksenia

```python
import sys
sys.setrecursionlimit(10**7)
n, m = map(int, input().split())
board = [list(input().strip()) for _ in range(n)]
dist = [[-1]*m for _ in range(n)]
visited = [[0]*m for _ in range(n)]
dirs = {'<': (0, -1), '>': (0, 1), '^': (-1, 0), 'v': (1, 0)}
def dfs(i, j):
    if board[i][j] == '#':
        return 0
    if visited[i][j] == 1:
        print(-1)
        sys.exit()
    if visited[i][j] == 2:
        return dist[i][j]

    visited[i][j] = 1
    dx, dy = dirs[board[i][j]]
    ni, nj = i + dx, j + dy

    dist[i][j] = 1 + dfs(ni, nj)
    visited[i][j] = 2
    return dist[i][j]
lengths = []
for i in range(n):
    for j in range(m):
        if board[i][j] != '#':
            lengths.append(dfs(i, j))
lengths.sort(reverse=True)
if len(lengths) == 0:
    print(0)
elif len(lengths) == 1:
    print(lengths[0] * 2)
else:
    print(lengths[0] + lengths[1])
```

### C. k-Tree 

```python
MOD = 10**9 + 7

n, k, d = map(int, input().split())

dp = [[0, 0] for _ in range(n + 1)]
dp[0][0] = 1  
for i in range(1, n + 1):
    for j in range(1, k + 1):
        if i - j >= 0:
            if j < d:
                dp[i][0] = (dp[i][0] + dp[i - j][0]) % MOD
                dp[i][1] = (dp[i][1] + dp[i - j][1]) % MOD
            else:
                dp[i][1] = (dp[i][1] + dp[i - j][0] + dp[i - j][1]) % MOD

print(dp[n][1] % MOD)
```

### D. Roads not only in Berland

```python
class DSU:
    def __init__(self, n):
        self.parent = list(range(n + 1))
    def find(self, x):
        if self.parent[x] != x:
            self.parent[x] = self.find(self.parent[x])
        return self.parent[x]
    def union(self, a, b):
        ra, rb = self.find(a), self.find(b)
        if ra != rb:
            self.parent[rb] = ra
            return True
        return False
n = int(input())
dsu = DSU(n)
extra = []  
for _ in range(n - 1):
    a, b = map(int, input().split())
    if not dsu.union(a, b):
        extra.append((a, b))
new_roads = []
for i in range(2, n + 1):
    if dsu.find(i) != dsu.find(1):
        new_roads.append((1, i))
        dsu.union(1, i)
print(len(extra))
for (a, b), (x, y) in zip(extra, new_roads):
    print(a, b, x, y)
```
