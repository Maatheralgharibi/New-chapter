### 1- Liar Liar Pants on Fire
  

```python
T = int(input())
for _ in range(T):
    n = int(input())
    a = list(map(int, input().split()))
    max_truth = 0
    for suspect in range(n):
        truth_count = sum(1 for i in range(n) if a[i] == suspect)
        max_truth = max(max_truth, truth_count)
    if n - max_truth >= n - 1:
        print("Definitely lying")
    else:
        print("Not Definitely Lying")
```

###  2- You Will Never Walk Alon


```python
n = int(input())
s, h = map(int, input().split())
found = False
for _ in range(n):
    a, b = map(int, input().split())
    if a == s and b == h:
        found = True
print("YES" if found else "NO")
```

### 3- Prime Time

```python
from collections import Counter

def factorize(n):
    factors = Counter()
    i = 2
    while i * i <= n:
        while n % i == 0:
            factors[i] += 1
            n //= i
        i += 1
    if n > 1:
        factors[n] += 1
    return factors

t = int(input())
for _ in range(t):
    a, b = map(int, input().split())
    fa = factorize(a)
    fb = factorize(b)
    all_primes = set(fa.keys()).union(fb.keys())
    ops = 0
    for p in all_primes:
        ops += abs(fa.get(p, 0) - fb.get(p, 0))
    print(ops)


    
```
