### 1- Liar Liar Pants on Fire
  

```python
def has_cycle(n, arr):
    visited = [0] * (n + 1)  # 0 = not visited, 1 = visiting, 2 = done

    def dfs(u):
        if visited[u] == 1:
            return True
        if visited[u] == 2 or arr[u] == u:
            return False
        visited[u] = 1
        if dfs(arr[u]):
            return True
        visited[u] = 2
        return False

    for i in range(1, n + 1):
        if visited[i] == 0:
            if dfs(i):
                return True
    return False


T = int(input())
for _ in range(T):
    n = int(input())
    a = [0] + list(map(int, input().split()))
    if has_cycle(n, a):
        print("Definitely Lying")
    else:
        print("Not Definitely Lying")
```

###  2- You Will Never Walk Alon


```python
n = int(input())
s, h = map(int, input().split())
start, end = min(s, h), max(s, h)

segments = []
for _ in range(n):
    a, b = map(int, input().split())
    l, r = min(a, b), max(a, b)
    if r < start or l > end:
        continue  
    segments.append((max(l, start), min(r, end)))  
segments.sort()
covered = start
for l, r in segments:
    if l > covered:
        print("NO")
        break
    covered = max(covered, r)
    if covered >= end:
        print("YES")
        break
else:
    print("NO")
```

### 3- Prime Time

```python
from collections import Counter

def factorize(n):
    factors = Counter()
    i = 2
    while i * i <= n:
        while n % i == 0:
            factors[i] += 1
            n //= i
        i += 1
    if n > 1:
        factors[n] += 1
    return factors

t = int(input())
for _ in range(t):
    a, b = map(int, input().split())
    fa = factorize(a)
    fb = factorize(b)
    all_primes = set(fa.keys()).union(fb.keys())
    ops = 0
    for p in all_primes:
        ops += abs(fa.get(p, 0) - fb.get(p, 0))
    print(ops)


    
```
